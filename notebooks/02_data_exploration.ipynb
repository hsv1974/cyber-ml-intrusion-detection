```python
import pandas as pd
from pathlib import Path

data_dir = Path("../data/raw")

train_file = data_dir / "UNSW_NB15_training-set.csv"
test_file = data_dir / "UNSW_NB15_testing-set.csv"

print("Training file exists:", train_file.exists())
print("Testing file exists:", test_file.exists())
```

```python
train_df = pd.read_csv(train_file)
test_df = pd.read_csv(test_file)

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)
```

```python
print("Training columns:")
print(train_df.columns.tolist())
```

```python
train_df.head()
```

```python
print("DATA TYPES")
print(train_df.dtypes)

print("\nMISSING VALUES")
missing = train_df.isnull().sum()
print(missing[missing > 0])

print("\nTOTAL MISSING VALUES:", train_df.isnull().sum().sum())
```

```python
print("LABEL COUNTS")
print(train_df["label"].value_counts())

print("\nLABEL PERCENTAGES")
print(train_df["label"].value_counts(normalize=True) * 100)
```

```python
print("ATTACK CATEGORIES")
print(train_df["attack_cat"].value_counts())
```

```python
categorical_cols = train_df.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = train_df.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nNumber of categorical columns:", len(categorical_cols))
print("Number of numerical columns:", len(numerical_cols))
```

```python
for col in ["proto", "service", "state"]:
    print(f"\n--- {col.upper()} ---")
    print("Unique values:", train_df[col].nunique())
    print(train_df[col].value_counts().head(15))
```

```python
print("Duplicate training rows:", train_df.duplicated().sum())
print("Duplicate testing rows:", test_df.duplicated().sum())
```

```python
print("Same columns:", list(train_df.columns) == list(test_df.columns))

print("\nTraining:", train_df.shape)
print("Testing:", test_df.shape)
```
